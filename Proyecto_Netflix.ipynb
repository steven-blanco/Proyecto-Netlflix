**Paso 1 . Carga de librerías y exploración de datos**


```python
#Importar librerias
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Cargar dataset
df = pd.read_csv("/content/drive/MyDrive/BI_Netflix_Exploded.csv")
```

```python
#Exploración del dataset
df.head(20)
```

Observamos dos cosas importantes en el dataset:
1.   Existen producciones repetidas en la lista cuyo única diferenciación es el dato registrado en la columna `listed_in`.
2.   Existen producciones que fueron colaboraciones entre distintos países. Se trabajará más adelante con ello.

```python
#Observación de conteo y formato en el df original
df.info()
```

```python
#Creación de los datasets
data_país = df.copy()
data_rating = df.copy()
```

```python
data_país.head(20)
```

```python
data_rating.head(20)
```

```python
#Eliminamos las columnas no necesarias para los datasets
data_país.drop(['director', 'cast', 'duration', 'date_added','listed_in'], axis = 1, inplace = True, errors= 'ignore')
data_país.reset_index(drop=True, inplace=True)
data_rating.drop(['director', 'cast', 'duration', 'date_added','listed_in','country'], axis = 1, inplace = True, errors= 'ignore')
data_rating.reset_index(drop=True, inplace=True)
```

Elimino las columnas que no necesito en mis datasets nuevos.

```python
#Verifico cambios
data_país.head(20)
```

```python
data_rating.head(20)
```

Notemos que con la eliminación de la columna `listed_in` en el data_rating, existen filas completamente duplicadas. Nos encargaremos más adelante.

```python
#Información de nuevos datasets
data_país.info()
data_rating.info()
```

Los dataframes creados contienen inicialmente la misma cantidad de filas que el dataset original.

```python
#Separación de países
data_país['country'] = data_país['country'].str.split(', ')
data_país = data_país.explode('country')
data_país.head(20)
```

```python
data_país.shape
```

El código para data_país logró separar en filas distintas las producciones en colaboración con distintos países para estandarizar el conteo. Notamos que:
1.   Existen filas con información completamente repetida al haber eliminado la columna `listed_in`
2.   Al separar en filas distintas las producciones con países en colaboración, se sumaron las entradas del dataset.





```python
#Verificamos duplicados
data_país.duplicated().sum()
```

```python
data_rating.duplicated().sum()
```

```python
#Eliminamos las filas completamente iguales
data_país = data_país.drop_duplicates(keep='first').reset_index(drop=True)
data_país.duplicated().sum()
```

```python
data_país.head(20)
```

```python
data_rating = data_rating.drop_duplicates(keep='first').reset_index(drop=True)
data_rating.duplicated().sum()
```

```python
print(data_país.shape)
print(data_rating.shape)
```

```python
#Observamos la proporción de los nulos en los datasets
data_país.info()
data_país.isna().sum()
data_país.isna().sum()/len(data_país)*100
```

```python
data_rating.info()
data_rating.isna().sum()
data_rating.isna().sum()/len(data_rating)*100
```

Las métricas de este proyecto se basarán en tres df distintos:
1. DF original para no perder información importante para las métricas de `type`y `listed_in`, independientemente de los valores de otras columnas.
2. data_país y data_rating para obtener métricas sin perder información por nulos en registros por columna clave en dichos df's.

```python
#Eliminamos las filas con valores nulos en las columnas seleccionadas en ambos sets.
data_rating.dropna(subset=['rating'], inplace=True)
data_rating.reset_index(drop=True, inplace=True)
data_país.dropna(subset=['country'], inplace=True)
data_país.reset_index(drop=True, inplace=True)
```

```python
#Comparación de cambios.
data_país.info()
data_país.isna().sum()
data_país.isna().sum()/len(data_país)*100
```

```python
data_rating.info()
data_rating.isna().sum()
data_rating.isna().sum()/len(data_rating)*100
```

```python
#Guardamos datasets
data_rating.to_csv('df_listed.csv', index=False)
data_país.to_csv('df_país.csv', index=False)
```

**Paso 2: Métricas clave**

```python
#Cantidad de producciones en el catálogo de Netflix
print(df['show_id'].nunique())
```

```python
#País con mayor cantidad de producciones
data_país.groupby('country')['show_id'].nunique().sort_values(ascending=False).head(1)
```

```python
#Porcentaje de implicación del país más popular en el catálogo
porcentaje = (
    data_país.groupby('country')['show_id'].nunique()
    / data_país['show_id'].nunique()
    * 100
)
porcentaje.sort_values(ascending=False).head(1)
```

```python
#Top 20 de países con menor cantidad de producciones
data_país.groupby('country')['show_id'].nunique().sort_values(ascending=True).head(20)
```

```python
#Cantidad de producciones que existen en el catálogo según tipo
df.groupby('type')['show_id'].nunique().sort_values(ascending=False).head()
```

```python
#Mayor género de producciones por tipo de producción
conteo = (
    df.groupby(['type', 'listed_in'])['show_id']
      .nunique()
      .reset_index(name='producciones')
)

resultado = conteo.loc[
    conteo.groupby('type')['producciones'].idxmax()
]

print(resultado)
```

```python
#Cantidad más popular de 'rating' clasificado por tipo de producción
resultado = (
    data_rating.groupby('type')['rating']
    .value_counts()
    .groupby(level=0)
    .head(1)
)
resultado
```
